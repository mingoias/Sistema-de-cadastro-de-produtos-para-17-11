```python
vetor = [3, 5, 4, 1, 6, 2]

comparacoes = 0
trocas = 0

inicio = 0
fim = len(vetor) - 1

pilha = [(inicio, fim)]

while pilha:

    inicio, fim = pilha.pop()

    if inicio < fim:

        pivo = vetor[fim]

        i = inicio - 1

        for j in range(inicio, fim):

            comparacoes += 1

            if vetor[j] <= pivo:

                i += 1

                if i != j:
                    vetor[i], vetor[j] = vetor[j], vetor[i]
                    trocas += 1

        if i + 1 != fim:
            vetor[i + 1], vetor[fim] = vetor[fim], vetor[i + 1]
            trocas += 1
            print('#########')
            print(vetor)

        posicao_pivo = i + 1

        if inicio < posicao_pivo - 1:
            pilha.append((inicio, posicao_pivo - 1))

        if posicao_pivo + 1 < fim:
            pilha.append((posicao_pivo + 1, fim))

print("Vetor ordenado:", vetor)
print("Comparações:", comparacoes)
print("Trocas:", trocas)
```

```python
import random

# Tamanhos dos vetores que serão testados
tamanhos = [10, 20, 1000]

for tamanho in tamanhos:

    vetor = [random.randint(1,1000) for i in range(tamanho)]

    # ==================================================
    # BUBBLE SORT
    # ==================================================

    bubble = vetor.copy()

    comparacoes_bubble = 0
    trocas_bubble = 0

    n = len(bubble)

    for i in range(n - 1):

        trocou = False

        for j in range(n - 1 - i):

            comparacoes_bubble += 1

            if bubble[j] > bubble[j + 1]:

                bubble[j], bubble[j + 1] = bubble[j + 1], bubble[j]

                trocas_bubble += 1

                trocou = True

        if trocou == False:
            break


    # ==================================================
    # QUICK SORT
    # ==================================================

    quick = vetor.copy()

    comparacoes_quick = 0
    trocas_quick = 0

    pilha = [(0, len(quick) - 1)]

    while pilha:

        inicio, fim = pilha.pop()

        if inicio < fim:

            # Escolhe o último elemento como pivô
            pivo = quick[fim]

            i = inicio - 1

            for j in range(inicio, fim):

                comparacoes_quick += 1

                if quick[j] <= pivo:

                    i += 1

                    if i != j:

                        quick[i], quick[j] = quick[j], quick[i]

                        trocas_quick += 1

            # Coloca o pivô na posição correta
            if i + 1 != fim:

                quick[i + 1], quick[fim] = quick[fim], quick[i + 1]

                trocas_quick += 1

            posicao_pivo = i + 1

            # Parte esquerda
            if inicio < posicao_pivo - 1:
                pilha.append((inicio, posicao_pivo - 1))

            # Parte direita
            if posicao_pivo + 1 < fim:
                pilha.append((posicao_pivo + 1, fim))


    # ==================================================
    # RESULTADOS
    # ==================================================

    print("\n========================================")
    print("VETOR COM", tamanho, "ELEMENTOS")
    print("========================================")

    # Mostra o vetor somente se ele for pequeno
    if tamanho <= 20:
        print("Vetor original:", vetor)

    print("\nBUBBLE SORT")
    print("Comparações:", comparacoes_bubble)
    print("Trocas:", trocas_bubble)
    print("Total de operações:", comparacoes_bubble + trocas_bubble)

    print("\nQUICK SORT")
    print("Comparações:", comparacoes_quick)
    print("Trocas:", trocas_quick)
    print("Total de operações:", comparacoes_quick + trocas_quick)
```

```python
valor = int(input("Digite o valor que deseja procurar: "))
tamanhos = [2, 10, 100]
resultados = []
for tamanho in tamanhos:
    matriz = []
    numero = 1
    for i in range(tamanho):
        linha = []
        for j in range(tamanho):
            linha.append(numero)
            numero += 1
        matriz.append(linha)

    comparacoes = 0
    encontrado = False
    linha_encontrada = -1
    coluna_encontrada = -1
    for i in range(tamanho):
        for j in range(tamanho):
            comparacoes += 1
            if matriz[i][j] == valor:
                encontrado = True
                linha_encontrada = i
                coluna_encontrada = j
                break
        if encontrado:
            break

    resultados.append([
        tamanho,
        tamanho * tamanho,
        encontrado,
        linha_encontrada,
        coluna_encontrada,
        comparacoes
    ])

print("\n")
print("==========================================================")
print("RESULTADO DA BUSCA")
print("==========================================================")

print(
    f"{'Matriz   '}"
    f"{'Elementos   '}"
    f"{'Encontrado   '}"
    f"{'Linha   '}"
    f"{'Coluna   '}"
    f"{'Comparações   '}"
)

print("==========================================================")

for resultado in resultados:

    tamanho = resultado[0]
    elementos = resultado[1]
    encontrado = resultado[2]
    linha = resultado[3]
    coluna = resultado[4]
    comparacoes = resultado[5]

    print(
      f"{tamanho} x {tamanho:<8}"
      f"{elementos:<12}"
      f"{str(encontrado):<12}"
      f"{linha:<10}"
      f"{coluna:<10}"
      f"{comparacoes:<15}"
    )

print("==========================================================")
```
